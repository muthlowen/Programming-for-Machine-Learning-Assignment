Question 1

Problem Description:
This program prompts the user to enter a positive odd integer greater than or equal to 5.
The input is validated, and the user is repeatedly asked until a valid value is provided.
Once a valid number n is entered, the program prints an X-shaped pattern using asterisks (*) inside an nxn grid.
Asterisks appear only on the two diagonals, and all other positions are spaces.

Assumptions:
The user may enter invalid input such as: even numbers, numbers less than 5, and non-numeric values.
The program must not crash on a invalid input.
The pattern must be generated dynamically based on user input.

Logic Explanation:
Use a while loop to continuously prompt the user.
Convert input to an integer using try-except to catch invalid input.
Validate: number >= 5, and number is odd.
Once valid input is received: use nested loops to print rows and columns. Print * if: row index equals column index (main diagonal), or row index + column index equals n-1 (secondary diagonal). Print a space otherwise.

```python
while True: #Continuously prompt the user until valid input is entered
  try:
    n = int(input("Please enter a positive odd integer that is greater or equal to 5: ")) #Ask user for input and convert to integer

    #Check if the number meets the conditions
    if n < 5: #Check if the number is lower than 5
      print("Error: Number must be at least 5.") #Reject if number is less than 5
    elif n % 2 == 0: #Check if number is even
      print("Error: Number must be odd.") #Reject if number is even
    else:
      break #Exit loop if input is valid
  except ValueError:
    print("Error: Please enter a valid integer.") #Handle non-numeric input

#Print the X-shaped pattern
#Outer loop iterate through each row
for row in range(n):
  #Inner loop iterate through each column in the current row
  for col in range(n):
    #Check if position lies on either diagonal
    if row == col:
      #Main digonal condition top-left to bottom-right
      print("*", end="")
    elif row + col == n - 1:
      #Secondary diagonal condition top-right to bottom-left
      print("*", end="")
    else:
      #Otherwise print space to maintain alignment
      print(" ", end="")
  print() #Move to next line after each row
```

Problem Description:
This program accepts a sentence from the user and identifies all unique palindromic words.
A palindromic word: reads the same forwards and backwards, is case-insensitive, and has at least 3 characters.
The program ignores punctuation and maintains the order of first appearance.

Assumptions:
The sentence may contain punctuation marks.
Uppercase and lowercase letters are treated as the same.
Duplicate palindromic words should appear only once in the output.

Logic Explanation:
Ask the user to enter a sentence.
Remove punctuation using the string.punctuation method.
Convert the sentence to lowercase.
Split the sentence into words.
For each word: check if its length is >= 3, check if it is equal to its reverse, and add it to the result list only if it is not already included.
Display: list of palindromic words, and total number found.

```python
import string

def find_palindromic_words(sentence):
  #Remove punctuation and convert to lowercase
  #str.maketrans creates a translation table to strip punctuation
  translate = str.maketrans("", "", string.punctuation)
  cleaned_sentance = sentence.translate(translate).lower()

  #Split sentence into words
  words = cleaned_sentance.split()
  palindromes = [] #List to store unique palindromic words

  #Iterate through each word in the sentence
  for word in words:
    #Check if word is >= 3 and is a palindrome
    #Word[::-1] reverses the string for comparison
    if len(word) >= 3 and word == word[::-1]:
      #Avoid duplicates by checking if already in list
      if word not in palindromes:
        palindromes.append(word)

  #Return the list of palindromic words found
  return palindromes

#Ask user for input
sentence = input("Please enter a sentence: ")

#Find palindromic words using the function
palindromic_words = find_palindromic_words(sentence)

#Display results
print("Palindromic words found:", palindromic_words)
print("Total number of palindromic words found:", len(palindromic_words))
```

Question 3

Problem Description:
This program simulates a simple shopping cart system for an online store.
The user eneters item prices repeatedly and types "done" when finished. The program validates all input and computes summary statistics.

Assumptions:
Prices must be positive numeric values.
"Done" can be entered in any combination of upper or lowercase.
The user may enter invalid values that must be handled properly.

Logic Explanation:
Create an empty list to store item prices.
Repeatedly prompt the user: if input is "done", stop input, and otherwise, attempt to convert input to a float.
Validate: value must be numeric, and value must be greater than zero.
After input is complete: calculate total items, total cost, average price, and identify highest and lowest prices.
Apply discount: if total >= Rs 1000, then apply 5% discount.
Display all monetary values to two decimal places.

```python
prices = [] #List to store all entered item prices

while True: #Continously prompt user until they type "done"
  user_input = input("Please enter item price (or type 'done' to finish): ").strip()
  #Prompt user for input and remove leading/trailing spaces with .strip()
  #Accepts either a numeric value or "done"

  #Check if user wants to stop entering prices
  if user_input.lower() == "done":
    #Convert input to lowercase
    #Break exits the loop
    break

  try:
    #Convert input to a floating point number
    price = float(user_input)

    #Validate that the price a positive
    if price <= 0:
      print("Error: Price must be a positive number.")
    else:
      #Add valid price to the list
      prices.append(price)
  except ValueError:
    #Handle invalid input
    print("Error: Please enter a valid number or 'done'.")

#Check if any items were entered
if prices:
  #Calculate summary statistics
  total_items = len(prices) #Count of items entered
  total_cost = sum(prices) #Sum of all prices
  average_price = total_cost / total_items #Mean price
  highest_price = max(prices) #Largest price entered
  lowest_price = min(prices) #Smallest price entered

  #Display results with formatted output
  print(f"Total items: {total_items}")
  print(f"Total cost: Rs {total_cost:.2f}")
  print(f"Average price: Rs {average_price:.2f}")
  print(f"Highest price: Rs {highest_price:.2f}")
  print(f"Lowest price: Rs {lowest_price:.2f}")

  #Apply discount rule of 5% off if total cost is >= 1000
  if total_cost >= 1000:
    discounted_total = total_cost * 0.95
    print(f"Discounted total (5% applied): Rs {discounted_total:.2f}")
  else:
    print("No discount applied.")
else:
  #Case when no valid items were entered
  print("No items entered.")
```
